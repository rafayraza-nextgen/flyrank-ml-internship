## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
# Every FlyRank client has more pages than the team has hours to review. Right now that review order is picked by gut feel and whoever's shouting loudest.
# I want to know whether a page's own behavior — traffic, position, how long it's been untouched — can tell us before a human looks at it whether it's worth reviewing first.
# This lane is attractive because the starter pipeline already ships a working baseline and label for it (is_declining_label, trend_direction),
# so I can spend my time on the judgment — what's a fair comparison, what's leakage, what's a defensible claim — instead of building data plumbing from zero.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# Decision: which pages get put in front of a content strategist this week for a refresh.
# Who acts on it: the client-facing strategist or account manager, who has limited hours and picks from a ranked list rather than reading every page.
# Cost of a wrong call, both directions:

# False positive (flagged as declining, isn't really): wastes a strategist's time on a page that didn't need attention — an hour or two lost, and it erodes trust in the tool the next time it makes a recommendation.
# False negative (a genuinely declining page never surfaces): the page keeps losing traffic silently until someone notices in a monthly report, by which point the client has already lost weeks of visibility and the fix is more expensive.
# Because a false negative is slower to detect and costlier to unwind, a good ranking should lean toward not missing decliners, even at the cost of a few extra false positives in the queue.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# Three numbers below make the case this lane has real signal to work with: how common decline is,
#  whether visibility (impressions) and staleness look different for declining vs stable pages, and how skewed the "worth reviewing" pool actually is.

import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"
print("Starter data found. You're ready.")

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

decline_rate = df["is_declining_label"].mean()
print(f"Declining rate across all pages: {decline_rate:.1%}")

stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).mean()
print(f"Share of pages that are stale AND still getting real traffic: {stale_visible:.1%}")

med_gap = (
    df.groupby("is_declining_label")["days_since_last_update"].median()
)
print("\nMedian days since last update, by decline label:")
print(med_gap.round(0).to_string())
```
