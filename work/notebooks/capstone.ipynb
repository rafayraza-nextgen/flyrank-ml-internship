## 1. Question

*The research question and the decision it supports.*

Question: How can content teams automatically identify which web pages need immediate updates to improve search visibility?
Decision it supports: This helps SEO managers decide which pages to prioritize for title and content updates to save time and increase traffic.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

We used a 90-day time window of web traffic from the FlyRank internship dataset. To maintain strict data privacy and safety, all personally identifiable information was removed. Fields like client_hash_id, exact domain names, page URLs, and raw search queries were completely excluded so no client can be identified.

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('../data/raw/content_refresh_anonymized.csv')
print(f"Data successfully loaded. Total rows: {len(df)}")
```

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

We selected safe metrics: impressions_90d, clicks_90d, and an engineered ctr (Click-Through Rate). The label was a binary flag for pages with >1000 impressions but a CTR under 2%. The baseline was a strict human rule looking for >1000 impressions and <50 clicks. We used a Decision Tree Classifier to mirror this logic. For validation, we used a Client-Grouped Split to ensure the model learned universal rules instead of memorizing specific client data.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GroupKFold
from sklearn.metrics import precision_score, recall_score, accuracy_score

# Feature Engineering: Calculate CTR
df['ctr'] = df['clicks_90d'] / df['impressions_90d']
df['ctr'] = df['ctr'].fillna(0)

# Define the target label (Quick Win = 1)
# Rule: More than 1000 impressions and CTR under 2%
df['is_quick_win'] = ((df['impressions_90d'] > 1000) & (df['ctr'] < 0.02)).astype(int)

# Define features and grouping
features = ['impressions_90d', 'clicks_90d', 'ctr']
X = df[features]
y = df['is_quick_win']

# Use the correct column name from your dataset for the grouping
groups = df['client_id']

# Set up Client-Grouped Split to prevent data leakage
gkf = GroupKFold(n_splits=5)
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

When compared against the rigid human baseline on the same test split, the Decision Tree model matched it perfectly. It achieved 100% accuracy, precision, and recall on the isolated test set. The model learned to rely heavily on the proportional Click-Through Rate (CTR) feature rather than just raw volume.

```python
# Create the Interpretable Decision Tree Model
model = DecisionTreeClassifier(max_depth=3, random_state=42)

# Train and test the model using GroupKFold
for train_idx, test_idx in gkf.split(X, y, groups):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Train the model
    model.fit(X_train, y_train)

    # Test the model
    predictions = model.predict(X_test)
    break # We only need to show one fold for the results

# Print the honest results
print(f"Accuracy:  {accuracy_score(y_test, predictions) * 100}%")
print(f"Precision: {precision_score(y_test, predictions) * 100}%")
print(f"Recall:    {recall_score(y_test, predictions) * 100}%")
```

## 5. Limitations

*What this work cannot claim.*

This tool is not fully autonomous and cannot replace human editorial judgment. Our model provides directional, decision-support insights based on observed and measured historical patterns. It cannot score brand new pages (under 90 days old) and its recommendations may become stale if search engines release major algorithm updates.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

Quick Win - Update Title (High Priority): Pages with high impressions but low CTR. A human should rewrite the title tag.

Protect - Monitor Traffic (Secondary Priority): Pages with high impressions and healthy clicks. Monitor for sudden drops.

The No-Go List: Legal pages, checkout pages, and contact pages must never be automated based on these signals.

```python
# Apply the model to all data
df['prediction'] = model.predict(X)

# Filter only the pages the model flagged as Quick Wins
quick_wins = df[df['prediction'] == 1].copy()

# Sort by highest impressions first
playbook = quick_wins.sort_values(by='impressions_90d', ascending=False)

print("Top SEO Quick Wins (Ranked Playbook):")
display(playbook[['impressions_90d', 'clicks_90d', 'ctr']].head())
```
